```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from crewai import LLM

llm = LLM(
    model="gemini/gemini-3.1-flash-lite-preview",
    temperature=0.1
)
llm.call("Who invented crew AI framework.")
```

```python
from crewai import Agent, Task, Crew

report_analyst = Agent(
    role="Business Report Analyst",
    goal="Analyze project updates and create concise executive summaries",
    backstory="An experienced business analyst who specializes in converting raw project notes into structured executive-level reports",
    verbose=True,
    llm=llm
)

project_update = """
Project deployment completed successfully in the staging environment.
API integration testing is still pending due to dependency on the external vendor team.
Initial performance testing shows good response time under moderate load.
Need final approval before production release next week.
"""

summary_task = Task(
    description=f"""Analyze the following project update and generate a professional executive summary.
    Highlight:
    - Current progress
    - Pending items
    - Risks or dependencies
    - Next steps

    Project Update:
    '''{project_update}'''
    """,
    agent=report_analyst,
    expected_output="A concise and professional executive project summary.",
)

crew = Crew(
    agents=[report_analyst],
    tasks=[summary_task],
    verbose=True
)

result = crew.kickoff()
print(result)
```
